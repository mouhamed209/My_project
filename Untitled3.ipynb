```python
import os, torch, ray, pandas as pd, numpy as np
import torch.nn as nn, torch.optim as optim, flwr as fl
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from collections import OrderedDict
import kagglehub

device = torch.device("cpu")
if ray.is_initialized(): ray.shutdown()
ray.init(num_cpus=2, ignore_reinit_error=True)
```

```python
def get_raw_mscad():
    path = kagglehub.dataset_download("drjamailalsawwa/mscad")
    csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
    df = pd.read_csv(os.path.join(path, csv_files[0]))
    return df, 'label' if 'label' in df.columns else df.columns[-1]

df_raw, label_col = get_raw_mscad()

# معالجة التصنيفات
le = LabelEncoder()
df_raw[label_col] = df_raw[label_col].astype(str).str.strip()
le.fit(df_raw[label_col])
num_classes = len(le.classes_)

# تقسيم البيانات: 90% للعملاء و 10% للخادم (Centralized Evaluation)
from sklearn.model_selection import train_test_split
df_clients, df_server = train_test_split(df_raw, test_size=0.1, stratify=df_raw[label_col])

# حساب أوزان الفئات
weights = compute_class_weight('balanced', classes=np.unique(le.transform(df_raw[label_col])), y=le.transform(df_raw[label_col]))
class_weights_tensor = torch.FloatTensor(weights).to(device)

def finalize(df, encoder, scaler=None):
    y = encoder.transform(df[label_col])
    X = df.drop(label_col, axis=1).select_dtypes(include=[np.number]).values
    if scaler is None:
        scaler = StandardScaler(); X = scaler.fit_transform(X)
    else: X = scaler.transform(X)
    X_lstm = X.reshape((X.shape[0], 1, X.shape[1]))
    ds = TensorDataset(torch.FloatTensor(X_lstm), torch.LongTensor(y))
    return DataLoader(ds, batch_size=64, shuffle=False), scaler, X.shape[1]

# تجهيز بيانات العملاء (Non-IID)
dfs = np.array_split(df_clients.sample(frac=1), 3)
train_loaders, global_scaler = [], None
for d in dfs:
    ldr, global_scaler, in_dim = finalize(d, le, scaler=global_scaler)
    train_loaders.append(ldr)

# تجهيز بيانات الخادم
server_loader, _, _ = finalize(df_server, le, scaler=global_scaler)
```

```python
class BiLSTMModel(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, 64, num_layers=1, batch_first=True, bidirectional=True)
        self.fc = nn.Linear(64 * 2, num_classes)
    def forward(self, x):
        out, _ = self.lstm(x)
        return self.fc(out[:, -1, :])

class FlowerClient(fl.client.NumPyClient):
    def __init__(self, model, loader):
        self.model, self.loader = model, loader
    def get_parameters(self, config):
        return [val.cpu().numpy() for _, val in self.model.state_dict().items()]
    def set_parameters(self, parameters):
        params_dict = zip(self.model.state_dict().keys(), parameters)
        state_dict = OrderedDict({k: torch.tensor(v) for k, v in params_dict})
        self.model.load_state_dict(state_dict, strict=True)
    def fit(self, parameters, config):
        self.set_parameters(parameters)
        criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)
        optimizer = optim.Adam(self.model.parameters(), lr=0.001)
        for _ in range(2):
            for data, target in self.loader:
                optimizer.zero_grad()
                criterion(self.model(data), target).backward(); optimizer.step()
        return self.get_parameters(config={}), len(self.loader.dataset), {}
    def evaluate(self, parameters, config):
        self.set_parameters(parameters)
        loss, correct, total = 0.0, 0, 0
        with torch.no_grad():
            for data, target in self.loader:
                outputs = self.model(data)
                loss += nn.CrossEntropyLoss()(outputs, target).item()
                correct += (torch.max(outputs, 1)[1] == target).sum().item()
                total += target.size(0)
        return float(loss) / len(self.loader), total, {"accuracy": float(correct / total)}
```

```python
# 1. تجميع مقاييس العملاء
def weighted_average(metrics):
    accuracies = [num_examples * m["accuracy"] for num_examples, m in metrics]
    examples = [num_examples for num_examples, _ in metrics]
    return {"accuracy": sum(accuracies) / sum(examples)}

# 2. تقييم النموذج العالمي في الخادم
def get_evaluate_fn(model, loader):
    def evaluate(server_round, parameters, config):
        params_dict = zip(model.state_dict().keys(), parameters)
        state_dict = OrderedDict({k: torch.tensor(v) for k, v in params_dict})
        model.load_state_dict(state_dict, strict=True)
        loss, correct, total = 0.0, 0, 0
        with torch.no_grad():
            for data, target in loader:
                outputs = model(data)
                loss += nn.CrossEntropyLoss()(outputs, target).item()
                correct += (torch.max(outputs, 1)[1] == target).sum().item()
                total += target.size(0)
        return float(loss) / len(loader), {"accuracy": float(correct / total)}
    return evaluate

# تشغيل
model_for_server = BiLSTMModel(in_dim, num_classes).to(device)
strategy = fl.server.strategy.FedAvg(
    evaluate_metrics_aggregation_fn=weighted_average,
    evaluate_fn=get_evaluate_fn(model_for_server, server_loader)
)

print("Starting Simulation...")
history = fl.simulation.start_simulation(
    client_fn=lambda cid: FlowerClient(BiLSTMModel(in_dim, num_classes).to(device), train_loaders[int(cid)]),
    num_clients=3,
    config=fl.server.ServerConfig(num_rounds=5),
    strategy=strategy,
    client_resources={"num_cpus": 1, "num_gpus": 0.0}
)
```

```python
import pickle

# 1. حفظ أوزان النموذج (عقل النموذج)
torch.save(model_for_server.state_dict(), 'bilstm_mscad_model.pth')

# 2. حفظ أداة التطبيع (ميزان البيانات)
with open('scaler.pkl', 'wb') as f:
    pickle.dump(global_scaler, f)

# 3. حفظ مشفر التصنيفات (قاموس الأسماء)
with open('label_encoder.pkl', 'wb') as f:
    pickle.dump(le, f)

print("✅ تم حفظ النموذج، الـ Scaler، والـ Encoder بنجاح!")
```

```python
def test_client_dataframe(client_index):
    # 1. تحميل الأدوات اللازمة
    with open('scaler.pkl', 'rb') as f:
        loaded_scaler = pickle.load(f)
    with open('label_encoder.pkl', 'rb') as f:
        loaded_le = pickle.load(f)

    test_model = BiLSTMModel(in_dim, num_classes)
    test_model.load_state_dict(torch.load('bilstm_mscad_model.pth'))
    test_model.eval()

    # 2. جلب بيانات العميل المختار من مصفوفة dfs التي أنشأناها سابقاً
    # dfs[0] هو العميل الأول، dfs[1] الثاني، وهكذا
    client_df = dfs[client_index].copy()

    # 3. تجهيز البيانات (عزل السمات عن الملصقات)
    X_raw = client_df.drop(label_col, axis=1).select_dtypes(include=[np.number]).values
    y_true = client_df[label_col].values

    # 4. التطبيع وتحويل الأبعاد للدفعة كاملة (Batch Processing)
    X_scaled = loaded_scaler.transform(X_raw)
    # نحولها لشكل (عدد الأسطر, 1, عدد السمات) ليقبلها الـ Bi-LSTM
    X_tensor = torch.FloatTensor(X_scaled).reshape((X_scaled.shape[0], 1, X_scaled.shape[1]))

    # 5. التنبؤ للبيانات بالكامل
    with torch.no_grad():
        outputs = test_model(X_tensor)
        predictions = torch.argmax(outputs, dim=1)

    # 6. تحويل الأرقام لأسماء تصنيفات
    y_pred = loaded_le.inverse_transform(predictions.numpy())

    # 7. عرض النتائج باحترافية
    print(f"📊 نتائج اختبار العميل رقم ({client_index + 1})")
    print(f"عدد العينات التي تم اختبارها: {len(client_df)}")
    print("-" * 40)

    from sklearn.metrics import classification_report, accuracy_score
    print(f"الدقة الإجمالية (Accuracy): {accuracy_score(y_true, y_pred):.2%}")
    print("\nتقرير التصنيف التفصيلي:")
    print(classification_report(y_true, y_pred))

# --- تشغيل الاختبار ---
# اختر العميل 0 أو 1 أو 2
test_client_dataframe(client_index=0)
```

```python
import torch
import pickle
import numpy as np

def predict_manual(sample_data):
    """
    هذه الدالة تأخذ بيانات سطر واحد وتخبرك بنوع الحركة (طبيعية أم هجوم)
    """
    # 1. تحميل الأدوات التي حفظناها سابقاً
    try:
        with open('scaler.pkl', 'rb') as f:
            loaded_scaler = pickle.load(f)
        with open('label_encoder.pkl', 'rb') as f:
            loaded_le = pickle.load(f)

        # 2. بناء هيكل النموذج وتحميل الأوزان الذكية
        # تأكد أن in_dim و num_classes معرفة من الخلايا السابقة
        test_model = BiLSTMModel(in_dim, num_classes)
        test_model.load_state_dict(torch.load('bilstm_mscad_model.pth'))
        test_model.eval()

        # 3. معالجة البيانات (Normalization & Reshape)
        sample_scaled = loaded_scaler.transform(sample_data)
        # البعد (1, 1, features) لأننا نختبر عينة واحدة في زمن واحد
        sample_tensor = torch.FloatTensor(sample_scaled).reshape((1, 1, in_dim))

        # 4. التنبؤ (Prediction)
        with torch.no_grad():
            output = test_model(sample_tensor)
            probabilities = torch.softmax(output, dim=1)
            prediction = torch.argmax(probabilities, dim=1)

        # 5. تحويل الرقم لاسم (مثلاً من 0 لـ Normal)
        predicted_label = loaded_le.inverse_transform(prediction.numpy())[0]
        confidence = torch.max(probabilities).item() * 100

        return predicted_label, confidence

    except FileNotFoundError:
        return "Error: لم نجد ملفات النموذج المحفوظة. تأكد من تشغيل خلية الحفظ أولاً.", 0

# --- الآن لنجرب الدالة على سطر "Normal" ---
# سنبحث عن سطر عشوائي تصنيفه Normal في بيانات الخادم
normal_data = df_server[df_server[label_col] == 'Normal'].iloc[5:6]

if not normal_data.empty:
    sample_row = normal_data.drop(label_col, axis=1).values
    result, score = predict_manual(sample_row)

    print(f"🔍 نتيجة فحص العينة:")
    print(f"-----------------------")
    print(f"✅ التصنيف الحقيقي: Normal")
    print(f"🤖 توقع النموذج: {result}")
    print(f"📊 نسبة الثقة: {score:.2f}%")
```
